## Protokol waktu dan pemakaian ulang

CPU satu thread; catat hardware, RAM, versi, threadpool dan identitas VM.
Empat skenario: satu peserta, batch 32, batch 128, dan seluruh holdout.
Single-query memakai 64 peserta test yang dipilih dengan seed tetap tanpa
melihat label. Semua jalur mendapat peserta yang sama. Ada 5 warm-up dan
30 blok pengukuran; urutan jalur dirotasi antarpeserta/blok.

Waktu prediction mencakup transformasi model serta decoding empat arah.
Loading model, training, pembacaan Drive/CSV, plotting, dan evaluasi statistik
dilakukan di luar timer. Rekam median, IQR, P95, waktu batch, throughput dan
waktu amortisasi per peserta. P95 waktu batch per peserta bukan P95 latency
single-query. Waktu hasil fit lama tidak diperlakukan sebagai inference.

Cache per model, prediksi, evaluasi dan **blok timing** memeriksa isi dan
konfigurasi. Timing juga memeriksa hardware/identitas boot VM. Terputus di VM
yang sama dapat dilanjutkan; VM berbeda memicu timing baru agar tidak mencampur
mesin. Nilai timing tetap bersyarat pada kondisi mesin yang dicatat.

Jika ingin pengukuran waktu baru, ubah `force_timing=True` saja. Jika ingin
memulihkan seluruh model, gunakan `force_models=True`. Perubahan output
tampilan notebook tidak membatalkan cache. Gunakan konfigurasi default untuk
hasil protokol penuh; pengaturan lebih kecil akan ditandai sebagai diagnostik.
